```python
import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("../data")
```

```python
with open(DATA_DIR / "config" / "config.json") as f:
    config = json.load(f)

entities = {}
for role in ("Prosumers", "Consumers", "Producers"):
    for name, spec in config[role].items():
        entities[name] = {"role": role, "type": spec["Type"]}

pd.DataFrame(entities).T
```
